```python
# Enumerate:  when you want to get both the index (position) and the value of each item in a sequence, such as a list, tuple, or string.

my_list = ['Monkey', 'Dog', 'Ferret']

list(enumerate(my_list))
```

```python
my_list = ['Monkey', 'Dog', 'Ferret']

list(enumerate(my_list, start = 5))
```

```python
my_list = ['Monkey', 'Dog', 'Ferret']

list(enumerate(my_list, 5))
```

```python
# Append

list_1 = [1, 2, 3]
another_value = 4

list_1.append(another_value)
```

```python
print(list_1)
```

```python
list_1 = [1, 2, 3]
list_2 = [4, 5, 6]

list_1.append(list_2)
```
