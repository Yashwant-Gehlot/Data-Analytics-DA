# Len

```python
my_list = [1, 2, 3, 'Hey', 6, 9, 31]

len(my_list)
```

```python
my_dict = {1: 'Hey', 2: 'Hello'}

len(my_dict)
```

```python
phone_numbers = ['7898595785', '741852963741852963', '789456123987412365']

len(phone_numbers)

for num in phone_numbers:
    if len(num) == 10:
        print(num)
```
