# Range

```python
range(start, stop, step)
```

```python
range(10)
```

```python
for num in range(10):
    print(num)
```

```python
for num in range(3, 10):
    print(num)
```

```python
for num in range(0, 100, 2):
    print(num)

```

```python
for num in range(0, 100, 2):
    if num % 3 == 0:
        print(num)
```
