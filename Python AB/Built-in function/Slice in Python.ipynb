# Slice

```python
my_list = [1, 2, 3, 4, 5, 6, 7]

my_list[0]
```

```python
my_list[:3]
```

```python
my_list[:7:2]
```

```python
my_slice = slice(0, 7, 2)
```

```python
my_list[my_slice]
```

```python
my_list = [1, 2, 3, 4, 5, 6, 7]
```

```python
my_slice = slice(-1, -9, -1)

my_list[my_slice]
```

```python
my_slice = my_list[::-1]

my_slice
```
