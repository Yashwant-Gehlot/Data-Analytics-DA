# Strip

```python
name = "             Yash         "

print(name)
```

```python
print(name.strip())
```

```python
name = "       I love my son        Yash         "

print(name)
```

```python
print(name.strip())
```

```python
name = ".................Yash..............."

print(name)
```

```python
print(name.strip('.'))
```

```python
my_list = ['0Strawberry','1Chocolate','2Vanilla','3RockyRoad']

new_list = [num.strip('123456789') for num in my_list]

print(new_list)
```

```python
my_list1 = ['0Strawberry','1Chocolate','2Vanilla','3RockyRoad']
new_list = []
for num in my_list1:
    new_list.append(num.strip('1234567890'))

print(new_list)
```
