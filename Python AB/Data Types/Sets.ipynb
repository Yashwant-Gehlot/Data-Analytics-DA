# Sets

```python
first_set = {1, 2, 3}

print(first_set)
```

```python
second_set = {1, 2, 'Hey!'}

print(second_set)
```

```python
third_set = {1, 2, 3, 1, 2, 3, 1, 2, 3}
print(third_set)
```

```python
forth_set = {(1, 2, 3), 1, 2, 3}
print(forth_set)
```

```python
fifth_set = {7,7,1,1,3,3, 'Carrot', 'Carrot'}

print(fifth_set)
```

```python
fifth_set.update(3)
```

```python
fifth_set.update([3,2,1])
print(fifth_set)
```

```python
fifth_set.discard(2)
print(fifth_set)
```

```python
fifth_set.discard(5)
print(fifth_set)
```

```python
fifth_set.remove(1)
print(fifth_set)
```

```python
fifth_set.remove(5)
print(fifth_set)
```

```python
set_1 = {1,2,3,4,5}
set_2 = {4,5,6,7,8}

print(set_1 | set_2) #it shows union
```

```python
print(set_1 & set_2) #it shows intersection
```

```python
print(set_1 - set_2)
```

```python
print(set_2 - set_1)
```

```python
print(set_2 ^ set_1)
```
