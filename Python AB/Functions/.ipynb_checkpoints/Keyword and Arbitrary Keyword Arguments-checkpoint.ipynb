# Keyword and Arbitrary Keyword Arguments

```python
def number_squared_function(number, power):
    print(number ** power)
```

```python
number_squared_function(2, 3)
```

```python
number_squared_function(power= 2, 3)
```

```python
number_squared_function(3, power= 2)
```

```python
# **kwargs

def number_kwargs(**kwargs):
    print(kwargs["Greetings"], kwargs["Name"])
```

```python
number_kwargs(Greetings = "Namaste!", Name = "Yashwant")
```

```python
def number_kwargs2(**Names):
    for key, value in Names.items():
        print(key + ': ' + value)
```

```python
number_kwargs2(Name = "Yash", Name2 = "Khushi", Name3 = "Ayush")
```

```python
my_dict = {"1": "Yash", "Johney" : "Hey", "3": "watermalon"}
```

```python
number_kwargs2(**my_dict)
```

```python
def wild_function(one, *two, **three):

    print(one)

    print(two[0], two[2])

    print(three.values())
```

```python
my_dict2 = {"1": "Yash", "Johney" : "Hey", "3": "watermalon"}

wild_function(27, *["Yash", 27, "NameSujal"], **my_dict2)
```
