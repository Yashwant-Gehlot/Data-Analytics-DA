# Function + Passing Arguments

```python
def first_function():
    print("we did it!")
```

```python
first_function()
```

```python
def second_function(name):
    print(name, "did it!")

second_function("Yashwant")
```

```python
def number_squared_func(number, power):
    print(number**power)

number_squared_func(3, 2)
```
