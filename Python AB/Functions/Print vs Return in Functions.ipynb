# Print vs Return in Functions

```python
def return_example():
    return 100
```

```python
return_example()
```

```python
type(return_example())
```

```python
def print_example():
    print(100)
```

```python
print_example()
```

```python
type(print_example())
```

```python
return_example() + 200
```

```python
print_example() + 200
```
