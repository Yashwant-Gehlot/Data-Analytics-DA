# Assignment Operators

```python
num = 5

num = num + 5

print(num)
```

```python
num += 5

print(num)
```

```python
num = 5
num -= 5

print(num)
```

```python
num = 5
num *= 5

print(num)
```

```python
num = 5
num /= 5

print(num)
```

```python
num = 5
num %= 5

print(num)
```

```python
num = 5
num **= 5

print(num)
```
