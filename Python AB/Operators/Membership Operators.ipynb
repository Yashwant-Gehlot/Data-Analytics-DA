# Membership Operators

```python
love_mems = 'I love membership operators!'

'love' in love_mems
```

```python
'hate' in love_mems
```

```python
my_list = [1, 2, 3, 4, 5]

num_wanted = 9

num_wanted in my_list
```

```python
my_list = [1, 2, 3, 4, 5]

num_wanted = 9

num_wanted not in my_list
```

```python
my_dict = {1: 'Hey!', 2: 'Hello!'}

1 in my_dict
```

```python
my_dict = {1: 'Hey!', 2: 'Hello!'}

1 in my_dict

'Hey!' in my_dict.values()
```
