```python
# Addition, Subtraction, Multiplication, Division
```

```python
def add(x, y):
    return x + y
    
def substract(x, y):
    return x - y
    
def multiply(x, y):
    return x * y
    
def divide(x, y):
    return x/y
```

```python
print("""Choose a math operation
Enter "A" for Addition
Enter "S" for Substraction
Enter "M" for Multiplication
Enter "D" for Divide
""")

while True:
    choice = input("Enter a choice (A, S, M, D): ")

    num_one = int(input("Enter First Number: "))

    num_two = int(input("Enter Second Number: "))

    if choice.lower() in ["a", "add", "addition"]:
        print(f"{num_one} + {num_two} = {add(num_one, num_two)}")
    elif choice.lower() in ["s", "sub", "substract"]:
        print(f"{num_one} - {num_two} = {substract(num_one, num_two)}")
    elif choice.lower() in ["m", "mul", "multiplication"]:
        print(f"{num_one} * {num_two} = {multiply(num_one, num_two)}")
    elif choice.lower() in ["d", "div", "division", "divide"]:
        print(f"{num_one} / {num_two} = {divide(num_one, num_two)}")
    else:
        print("Please enter an valid input (A, S, M, D)")

    next_calculation = input("Do you want to do another calculation? (yes,no): ")
    if next_calculation.lower() in ['no','n','nope']:
        break
    else:
        continue
    

```
