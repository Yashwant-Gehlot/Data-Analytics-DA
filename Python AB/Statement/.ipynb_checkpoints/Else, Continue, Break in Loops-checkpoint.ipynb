# Break, Continue, Else, Pass in Loops

```python
# Break

counter = 0

while counter <= 10:
    print(counter)
    if counter == 6:
        break
    counter += 1
```

```python
integers = [1, 2, 3, 4, 5]

for num in integers:
    power_of_num = num ** num
    if power_of_num > 3000:
        break
    print(power_of_num)
```

```python
number = 0

while number < 5:
    number += 1
    if number == 3:
        continue
    print(number)
```

```python
integers = [1, 2, 3, 4, 5, 6, 7, 8]

for num in integers:
    power_of_num = num ** num
    if 200 <= power_of_num <= 3000:
        continue
    print(num, power_of_num)
```

```python
counter = 0

while counter <= 10:
    print(counter)
    if counter == 6.5:
        break
    counter += 1
else:
    print('No Breaks Found')
```

```python
integers = [1, 2, 3, 4, 5]

for num in integers:
    power_of_num = num ** num
    if power_of_num > 4000:
        break
    print(power_of_num)
else:
     power_of_num = num ** num
     print(power_of_num)
```
