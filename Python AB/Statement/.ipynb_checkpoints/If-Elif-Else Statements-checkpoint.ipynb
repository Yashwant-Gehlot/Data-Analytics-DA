# If-Elif-Else Statements

```python
if 100 > 10:
    print('It worked!')
```

```python
if 100 < 10:
    print('It worked!')
```

```python
if 100 < 10:
    print('It worked!')
else:
    print('It did not work!')
```

```python
invitee_list = ['Josh Hibbar', 'John Klows', 'Yashwant Gehlot', 'Yash Jain']

if 'Yashwant Gehlot' in invitee_list:
    print('Let them in')
else:
    print("Don't let them in...")
```

```python
price = 100

if price < 10:
    print('The price of this item is cheap.')
elif price < 40:
    print('The price of this item is not very expensive.')
elif price < 100:
    print('The price of this item is somewhat exensive')
else: 
    print('The price of this item is very expensive')
```

```python
# Elif


price = 75

if price < 10:
    print('The price is ' + str(price) + ' dollars. This is cheap')
elif price < 40:
    print('The price is ' + str(price) + ' dollars. This is not very expensive')
elif price < 100:
    print('The price is ' + str(price) + ' dollars. This is somewhat expensive')
else:
    print('The price is ' + str(price) + ' dollars. This is expensive')
```

```python
small_sale = []
medium_sale = []
large_sale = []
```

```python
sale = 1000
if sale < 20:
    small_sale.append(sale)
elif sale < 100:
    medium_sale.append(sale)
else:
    large_sale.append(sale)
print('Small Sale Number: ', small_sale)
print('Medium Sale Number: ', medium_sale)
print('Large Sale Number: ', large_sale)
```
