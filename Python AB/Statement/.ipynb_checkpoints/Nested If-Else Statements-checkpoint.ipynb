# Nested If - Else Statements

```python
grade = 84

if grade >= 90:
    if grade >= 97:
        print(str(grade) + ' = A+')
    elif grade >= 93:
        print(str(grade) + ' = A')
    else:
        print(str(grade) + ' = A-')
elif grade >= 80:
    if grade >= 87:
        print(str(grade) + ' = B+')
    elif grade >= 83:
        print(str(grade) + ' = B')
    else:
        print(str(grade) + ' = B-')

```
