# Nested While Loops

```python
number = 0

while number <= 10:
    print(number)
    inner_number = 0
    while inner_number < 5:
        print(inner_number)
        if inner_number == 2:
            break
        inner_number += 1
    else:
        print('Inner While loop ended')
    number +=1
else:
    print('Outer While loop ended')
```
