```python
sale_amount = [1, 10, 75.75, 100, 1000, 80, 45]

small_sale = []
medium_sale = []
large_sale = []

for sale in sale_amount:
    if sale < 10:
        small_sale.append(sale)
    elif sale < 75:
        medium_sale.append(sale)
    else:
        large_sale.append(sale)

print('small Sale: ', small_sale)
print('Medium Sale: ', medium_sale)
print('Large Sale: ', large_sale)
```

```python
flavor_ice_cream = ['Mint Chocolate Chip', 'Rocky Road', 'Pinapple Cream']

i = 0

for flavor in flavor_ice_cream:
    print(i, flavor)
    i += 1
```

```python
for counter, flavor in enumerate(flavor_ice_cream):
    print(counter, flavor)
```

```python
customer_dictionary = {'Name' : 'Yashwant Smith', 'Amount Spent': 76, 'items Purchased': ['Shovel', 'Dirt', 'Pot']}

for item in customer_dictionary:
    print(item)
```

```python
for item in customer_dictionary.values():
    print(item)
```

```python
for item in customer_dictionary.keys():
    print(item)
```

```python
for item in customer_dictionary.items():
    print(item)
```

```python
for key, value in customer_dictionary.items():
    print(key, '-->', value)
```

```python
for key, value in customer_dictionary.items():
    print(key, value)
```
